# Local Databricks Notebook
Welcome to your local notebook! This works exactly like a notebook in the Databricks UI.

```python
import os
from dotenv import load_dotenv
from databricks.connect import DatabricksSession
from IPython.core.magic import register_cell_magic

# 1. Connect to Databricks Serverless
load_dotenv()
spark = DatabricksSession.builder.getOrCreate()

# 2. Register a custom %%sql magic command
@register_cell_magic
def sql(line, cell):
    from IPython.display import display
    # We convert to Pandas so Jupyter renders the beautiful HTML table instantly!
    df = spark.sql(cell)
    display(df.toPandas())
    return None

print("✅ Connected! You can now use regular Python cells, or start a cell with %%sql")
```

### Writing SQL
Because we registered the magic above, you can now write pure SQL by putting `%%sql` at the very top of any cell!

```python
%%sql
SELECT * FROM dltprac.source.customers 
ORDER BY customer_id DESC
LIMIT 5
```

### Writing Python
And you can mix and match by using standard PySpark in the next cell:

```python
# Standard PySpark Python cell
df = spark.sql("SELECT * FROM dltprac.source.customers LIMIT 5")

# Use .toPandas() to render the table beautifully in VS Code!
df.toPandas()
```
